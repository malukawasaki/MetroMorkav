```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import requests
```

```python


DATASET_ID = "y2qv-fytt"
COLUMNS_URL = f"https://data.ny.gov/api/views/{DATASET_ID}/columns.json"
```

```python
BASE = f"https://data.ny.gov/resource/{DATASET_ID}.json"

WHERE = "year=2025 AND month=11 AND day_of_week='Monday'"  # change these
LIMIT = 50000  # start small; increase later

params = {
    "$where": WHERE,
    "$limit": LIMIT
}

df = pd.DataFrame(requests.get(BASE, params=params).json())
df.shape, df.head()
```

```python
# EDIT these after looking at df.columns:
HOUR_COL = "hour_of_day"
ORIGIN_COL = "origin_station_complex_id"
DEST_COL = "destination_station_complex_id"

od = df[[HOUR_COL, ORIGIN_COL, DEST_COL]].rename(
    columns={HOUR_COL:"hour", ORIGIN_COL:"origin", DEST_COL:"dest"}
).copy()

od["hour"] = od["hour"].astype(int)

od.head()
```
